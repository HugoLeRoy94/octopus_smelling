### Introduction

Here, we are presenting how to design an optimal array of receptors. 
Each receptor is made of $k_\text{sub}$ sub-units among $n_\text{unit}$ possible units.
We denote $\bm{r} = (r_0 \cdots r_{k_\text{sub}}) \in \mathbb{R}^{n_\text{unit} \times k_\text{sub}}$ the identity of a receptor characterized by the identity of each of the units that composes it. 
The response of a receptor $\bm{r}$ to a ligand $\ell$ at concentration $c$ can be written using the MWC model for ion channel opening:
$$
p_o(c,\ell) = \frac{\prod_{i\in \bm{r}} (1+c/K_o^{(i,\ell)})}{\prod_{i\in \bm{r}} (1+c/K_o^{(i,\ell)}) + e^{-\epsilon}\prod_{i\in \bm{r}}(1+c/K_c^{(i,\ell)})}
$$
Where the $K$'s are called the affinities and defined by:
$$
K_o^{(i,\ell)} = \exp\left[ \Delta E_o^{i,\ell} \right]
$$
$$
K_c^{(i,\ell)} = \exp\left[ \Delta E_c^{i,\ell} \right]
$$
Each affinity correspond to the response of a sub-unit to a random ligand. Each sub-unit has a different chemical properties, therefore the energy of a random ligand is drawn from a Normal distribution probability that depends on the specific sub-unit.
$$
P(E_\alpha^{(i,\ell)}) = \mathcal{N}(\mu_\alpha^i,\sigma_\alpha^i)
$$
Where $\alpha \in \{ o, c\}$.
Epsilon is the energy difference between the closed and opened state of the channel when there are no ligands, it is thus constant for each sub-unit.
The goal is to optimize the mean and standard deviation of the probability distribution of each individual sub-unit in order to maximize the information between a ligand's identity and its concentration.

Given an array of $N$ receptors, we write the activity of this array as $\mathcal{A} = (a_0,a_1, \cdots , a_N)$, and compute the mutual information between the activity of the array and ligand's identity and concentration:
$$
I(\mathcal{A};(c,\ell)) = H(\mathcal{A}) - H(\mathcal{A} | (c,\ell)),
$$
here $\ell$ represents the ensemble of affinities of the array 
Because the mapping between affinities, and activity is deterministic noiseless: $H(A | (c,\ell)) = 0$.
Thus we want to maximize $H(\mathcal{A}) = \sum p(\mathcal{A}) \log[p(\mathcal{A})]$.
To do so, we write:
$$
\begin{aligned}
p(\mathcal{A}) &= \sum_{i\in\bm{r}} \int \int \int \text{d}\Delta E_i^c p(\Delta E_i^c) \text{d} \Delta E_i^o p(\Delta E_i^o) \text{d}\epsilon p(\epsilon) p(\mathcal{A} | \{\Delta E_i^c, \Delta E_i^o, \epsilon_i \}_i) \\
& = \sum_{i \in \bm{r}} \int \text{d}\theta_i p(\theta_i) p(\mathcal{A}|\{\theta_i\}_i)
\end{aligned}
$$
Where $\theta_i = (\Delta E_i^c, \Delta E_i^o, \epsilon_i)$.
Now writing how this probability distribution vary with the mean or std deviation:
$$
\frac{\partial p(\mathcal{A})}{\partial w_i} = \int \text{d}\theta_i \left[ p(\mathcal{A} | \theta_i) \frac{\partial p(\theta_i)}{\partial w_i} + p(\theta_i) \frac{\partial p(\mathcal{A} | \theta_i)}{\partial \theta_i} \right]
$$
where $w_i$ could be $\mu_i^\alpha$ or $\sigma_i^\alpha$, and $\alpha \in \{o,c\}$.
The first term, is easy to compute because $p(\theta_i)$ is just a normal distribution, so it's easy to compute how it varies with the mean and std deviation.
